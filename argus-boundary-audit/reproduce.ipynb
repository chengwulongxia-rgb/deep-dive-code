本 Notebook 必須在能使用本機 Docker daemon 的電腦執行；不是一般 Colab/GPU Notebook。Build 需公開 PyPI 網路，測試 runtime 斷網。
先將 kernel 工作目錄設為本專案目錄；實際呼叫相同 reproduce.py，沒有憑證或預存執行輸出。

```python
from pathlib import Path
import subprocess, sys
project = Path.cwd().resolve()
assert (project / "reproduce.py").is_file(), "Set cwd to argus-boundary-audit first"
subprocess.run([sys.executable, "-B", str(project / "reproduce.py")], cwd=project, check=True, timeout=1800)
```

可於同一 runner 加上 --source 指定固定 SHA 的乾淨上游 checkout，或 --output 指定全新結果目錄。預設安全 fetch 固定 SHA 至 scratch。結果 manifest 與完整 stdout/stderr 保存在 results/run-*。
